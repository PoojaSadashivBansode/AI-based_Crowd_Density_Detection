```python
# 🎥 View Processed Video
from IPython.display import HTML
from base64 import b64encode

def play_video(path):
    if not os.path.exists(path):
        return print(f'❌ {path} not found')
    mp4 = open(path, 'rb').read()
    data_url = 'data:video/mp4;base64,' + b64encode(mp4).decode()
    return HTML(f'<video width=800 controls><source src="{data_url}" type="video/mp4"></video>')

play_video('output.mp4')
```

```python
# 📊 View Statistics & Logs
import pandas as pd

if os.path.exists('crowd_data.csv'):
    df = pd.read_csv('crowd_data.csv')
    print('📊 First 20 frames:')
    display(df.head(20))
    print(f'\n📈 Summary:')
    print(f'  • Frames: {len(df)}')
    print(f'  • YOLO: {(df["Mode"] == "YOLO").sum()}')
    print(f'  • CSRNet: {(df["Mode"] == "CSRNet").sum()}')
    print(f'  • Alerts: {(df["Alert"] == "YES").sum()}')
    print(f'  • Avg Count: {df["Count"].mean():.1f}')
    print(f'  • Peak: {df["Count"].max()}')
else:
    print('❌ crowd_data.csv not found')
```
